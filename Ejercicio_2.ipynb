#**Ejercicio 2: Dosificación de Fertilizante Líquido**


**Problema:** Un sistema de riego por goteo requiere aplicar una dosis de fertilizante por hectárea. El usuario ingresa las hectáreas totales y la dosis requerida (litros/hectárea).


**Datos:** El fertilizante se diluye en agua en una relación 1:500 (1 litro de producto por 500 de agua).


**Prueba:** 5 hectáreas, Dosis 2 L/ha. Salida: 10L de producto y 5000L de agua total.

````python
# 1. Entrada de datos
# Se solicita el área total y la dosis por hectárea [cite: 25]
hectareas = float(input("Ingrese las hectáreas totales: "))
dosis_l_ha = float(input("Ingrese la dosis requerida (L/ha): "))

# 2. Procesos y Conversiones
# Cálculo del fertilizante puro necesario
litros_producto = hectareas * dosis_l_ha

# Cálculo del agua necesaria según relación 1:500 (1L producto -> 500L agua)
litros_agua = litros_producto * 500

# 3. Salida de Resultados
print("-" * 30)
print(f"Resultado para {hectareas} hectáreas:")
print(f"Producto total: {litros_producto:.2f} L")
print(f"Agua para dilución: {litros_agua:.2f} L")
print("-" * 30)
````
