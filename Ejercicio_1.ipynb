#**Ejercicio 1: Capacidad de Silos de Granos**

**Problema:** Se requiere calcular cuántas toneladas de maíz puede almacenar un silo cilíndrico. El usuario ingresa el radio y la altura del silo en metros.


**Datos:** Densidad del maíz ≈0.75 ton/m3
. Volumen V=π*r2*h


**Prueba:** Radio = 3m, Altura = 10m. Salida ≈212.06
 toneladas.


```python
import math

def calcular_capacidad_silo():
    print("--- Calculadora de Capacidad de Silos (Maíz) ---")

    try:
        # Entrada de datos
        radio = float(input("Ingresa el radio del silo en metros: "))
        altura = float(input("Ingresa la altura del silo en metros: "))

        if radio <= 0 or altura <= 0:
            print("Error: Las dimensiones deben ser mayores a cero.")
            return

        # Constantes
        DENSIDAD_MAIZ = 0.75  # ton/m3

        # Fórmulas
        # Volumen = π * r² * h
        volumen = math.pi * (radio ** 2) * altura

        # Masa = Volumen * Densidad
        toneladas = volumen * DENSIDAD_MAIZ

        # Resultados
        print("-" * 40)
        print(f"Resultados para un silo de {radio}m de radio y {altura}m de altura:")
        print(f"  > Volumen total: {volumen:.2f} m³")
        print(f"  > Capacidad de almacenamiento: {toneladas:.2f} toneladas de maíz")
        print("-" * 40)

    except ValueError:
        print("Error: Por favor, ingresa solo valores numéricos.")

# Ejecutar la función
if __name__ == "__main__":
    calcular_capacidad_silo()
```

Aquí tienes un análisis línea por línea del código proporcionado:

*   `import math`: Esta línea importa el módulo `math`, que es necesario para utilizar la constante `math.pi` (el valor de Pi) en los cálculos.

*   `def calcular_capacidad_silo():`: Define una función llamada `calcular_capacidad_silo()`. Todo el código para calcular la capacidad del silo está encapsulado dentro de esta función.

*   `print("--- Calculadora de Capacidad de Silos (Maíz) ---")`: Muestra un mensaje de bienvenida al usuario cuando se ejecuta la función.

*   `try:`: Inicia un bloque `try-except`, que se utiliza para manejar posibles errores, como cuando el usuario ingresa texto en lugar de números.

*   `radio = float(input("Ingresa el radio del silo en metros: "))`: Solicita al usuario que ingrese el radio del silo en metros y convierte la entrada (que es una cadena de texto) a un número flotante. Este valor se almacena en la variable `radio`.

*   `altura = float(input("Ingresa la altura del silo en metros: "))`: De manera similar, solicita la altura del silo en metros y la convierte a un número flotante, almacenándola en la variable `altura`.

*   `if radio <= 0 or altura <= 0:`: Comprueba si el radio o la altura ingresados son menores o iguales a cero. Si alguna de estas condiciones es verdadera, las dimensiones no son válidas.

*   `print("Error: Las dimensiones deben ser mayores a cero.")`: Si las dimensiones no son válidas, imprime un mensaje de error.

*   `return`: Termina la ejecución de la función si se detecta un error en las dimensiones.

*   `DENSIDAD_MAIZ = 0.75  # ton/m3`: Define una constante `DENSIDAD_MAIZ` con el valor 0.75, que representa la densidad del maíz en toneladas por metro cúbico.

*   `volumen = math.pi * (radio ** 2) * altura`: Calcula el volumen del silo cilíndrico utilizando la fórmula V = π * r² * h. `math.pi` proporciona el valor de Pi, `radio ** 2` calcula el radio al cuadrado y `* altura` multiplica por la altura.

*   `toneladas = volumen * DENSIDAD_MAIZ`: Calcula la capacidad de almacenamiento en toneladas multiplicando el volumen calculado por la densidad del maíz.

*   `print("-" * 40)`: Imprime una línea de guiones para formatear la salida y hacerla más legible.

*   `print(f"Resultados para un silo de {radio}m de radio y {altura}m de altura:")`: Muestra un encabezado con el radio y la altura que el usuario ingresó.

*   `print(f"  > Volumen total: {volumen:.2f} m³")`: Imprime el volumen total del silo, formateado para mostrar dos decimales.

*   `print(f"  > Capacidad de almacenamiento: {toneladas:.2f} toneladas de maíz")`: Imprime la capacidad de almacenamiento en toneladas, también formateada a dos decimales.

*   `print("-" * 40)`: Imprime otra línea de guiones para el formato.

*   `except ValueError:`: Si ocurre un `ValueError` dentro del bloque `try` (por ejemplo, si el usuario ingresa texto en `input()` en lugar de números), el control pasa a este bloque.

*   `print("Error: Por favor, ingresa solo valores numéricos.")`: Imprime un mensaje de error si el usuario ingresó un valor no numérico.

*   `if __name__ == "__main__":`: Esta es una construcción estándar en Python. Significa que el código dentro de este bloque solo se ejecutará cuando el script se ejecute directamente (no cuando se importe como un módulo en otro script).

*   `calcular_capacidad_silo()`: Llama a la función `calcular_capacidad_silo()` para iniciar el programa cuando el script se ejecuta directamente.
