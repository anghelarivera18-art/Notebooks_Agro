#**Ejercicio 3: Rendimiento de Pulpa en Frutas**


**Problema:** Una planta procesadora necesita saber el porcentaje de aprovechamiento. Se ingresa el peso total de la fruta comprada y el peso de los residuos (cáscaras y semillas).

**Datos:** Pulpa=Peso Total−Residuos
. %Rendimiento=(Pulpa/Peso Total)*100
.

**Prueba:** Peso Total = 500kg, Residuos = 150kg. Salida: 350kg de pulpa (70%).

```python
peso_total = float(input("Peso total de la fruta comprada (kg): "))
peso_residuos = float(input("Peso de los residuos -cáscaras/semillas- (kg): "))

pulpa = peso_total - peso_residuos

rendimiento = (pulpa / peso_total) * 100

print("-" * 35)
print(f"Masa de pulpa obtenida: {pulpa:.2f} kg")
print(f"Porcentaje de rendimiento: {rendimiento:.1f}%")
print("-" * 35)
```

Aquí te explico el código línea por línea:

*   `peso_total = float(input("Peso total de la fruta comprada (kg): "))`:
    *   Esta línea solicita al usuario que ingrese el peso total de la fruta comprada en kilogramos.
    *   La función `input()` lee la entrada del usuario como una cadena de texto.
    *   `float()` convierte esa cadena de texto en un número decimal (flotante) para poder realizar cálculos matemáticos.
    *   El valor se guarda en la variable `peso_total`.

*   `peso_residuos = float(input("Peso de los residuos -cáscaras/semillas- (kg): "))`:
    *   De manera similar a la línea anterior, esta línea pide al usuario que ingrese el peso de los residuos (cáscaras y semillas) en kilogramos.
    *   La entrada se convierte a un número flotante y se almacena en la variable `peso_residuos`.

*   `pulpa = peso_total - peso_residuos`:
    *   Esta línea realiza el cálculo para obtener la masa de la pulpa.
    *   Resta el `peso_residuos` del `peso_total`.
    *   El resultado (la masa de pulpa) se asigna a la variable `pulpa`.

*   `rendimiento = (pulpa / peso_total) * 100`:
    *   Aquí se calcula el porcentaje de rendimiento.
    *   Se divide la `pulpa` entre el `peso_total` para obtener la proporción de pulpa.
    *   Luego, se multiplica por `100` para expresar esa proporción como un porcentaje.
    *   El valor del porcentaje se guarda en la variable `rendimiento`.

*   `print("-" * 35)`:
    *   Esta línea imprime una línea de 35 guiones (`-`) para crear un separador visual en la salida, haciendo que los resultados sean más fáciles de leer.

*   `print(f"Masa de pulpa obtenida: {pulpa:.2f} kg")`:
    *   Esta línea imprime la masa de pulpa obtenida.
    *   Usa una f-string (cadena de formato) para insertar el valor de la variable `pulpa` directamente en el texto.
    *   `:.2f` formatea el número `pulpa` para que se muestren solo dos decimales.

*   `print(f"Porcentaje de rendimiento: {rendimiento:.1f}%")`:
    *   Similar a la línea anterior, imprime el porcentaje de rendimiento.
    *   `:.1f` formatea el número `rendimiento` para que se muestre con un solo decimal.

*   `print("-" * 35)`:
    *   Imprime otra línea de 35 guiones para cerrar el bloque de resultados.
